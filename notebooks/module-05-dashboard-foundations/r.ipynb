### The filter logic the app's callback will run

Given a province, return the filtered frame the chart draws from — this is the function `apps/shiny_app.R` calls on every dropdown change.

````r
filter_by_province <- function(frame, prov) {
  frame |> filter(province == prov) |> arrange(date)
}

sample <- filter_by_province(df, df$province[1])
ggplot(sample, aes(date, value)) +
  geom_line(color = "#1D6E73") +
  labs(title = paste("Preview:", df$province[1]))
````

### The app

```r
# apps/shiny_app.R  (excerpt — full file at the repo root)
library(shiny)
library(ggplot2)

df <- read.csv("data/processed/track2_dataset.csv")

ui <- fluidPage(
  titlePanel("Regional Analytic Hub, Prototype"),
  sidebarLayout(
    sidebarPanel(selectInput("province", "Province", sort(unique(df$province)))),
    mainPanel(plotOutput("trend"))
  )
)

server <- function(input, output) {
  output$trend <- renderPlot({
    ggplot(subset(df, province == input$province), aes(date, value)) +
      geom_line(color = "#1D6E73")
  })
}

shinyApp(ui, server)
```

Run with `Rscript -e "shiny::runApp('apps/shiny_app.R')"` from the repo root.
