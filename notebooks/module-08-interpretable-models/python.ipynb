```python
import pandas as pd

DATA_PATH = "../../data/processed/track2_dataset.csv"
df = pd.read_csv(DATA_PATH, parse_dates=["date"])
print(f"{len(df):,} rows · {df['district'].nunique()} districts · "
      f"{df['date'].min().date()} to {df['date'].max().date()}")
df.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report

X = df[["feature_1", "feature_2"]]
y = df["outcome"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42, stratify=y)

model = LogisticRegression().fit(X_train, y_train)
preds = model.predict(X_test)
print(classification_report(y_test, preds))

coefs = pd.Series(model.coef_[0], index=X.columns).sort_values()
print("\nCoefficients (log-odds scale):\n", coefs)
```

**Plain-language summary (fill in after running):** the model correctly classifies about _[accuracy]_ of held-out cases. `feature_1` moves the odds of a positive outcome _[up/down]_ per unit increase; `feature_2` _[up/down]_.
